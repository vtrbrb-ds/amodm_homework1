# Introduction to Sets


```python
def average(array):
    return sum(set(array)) / len(set(array))


if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

# Symmetric Difference


```python
if __name__ == '__main__':
    m = int(input())
    set_m = set(map(int, input().split()))

    n = int(input())
    set_n = set(map(int, input().split()))

    result = sorted(set_m.symmetric_difference(set_n))

    for value in result:
        print(value)
```

# No Idea!


```python
if __name__ == '__main__':
    n, m = map(int, input().split())
    array = list(map(int, input().split()))

    liked = set(map(int, input().split()))
    disliked = set(map(int, input().split()))

    happiness = 0

    for number in array:
        if number in liked:
            happiness += 1
        elif number in disliked:
            happiness -= 1

    print(happiness)
```

# Set .add()


```python
if __name__ == '__main__':
    n = int(input())
    countries = set()

    for _ in range(n):
        countries.add(input().strip())

    print(len(countries))
```

# Set .discard(), .remove() & .pop()


```python
if __name__ == '__main__':
    n = int(input())
    s = set(map(int, input().split()))
    commands = int(input())

    for _ in range(commands):
        operation = input().split()

        if operation[0] == 'pop':
            s.pop()
        elif operation[0] == 'remove':
            s.remove(int(operation[1]))
        elif operation[0] == 'discard':
            s.discard(int(operation[1]))

    print(sum(s))
```

# Set Mutations


```python
if __name__ == '__main__':
    n = int(input())
    a = set(map(int, input().split()))

    operations = int(input())

    for _ in range(operations):
        command, size = input().split()
        other = set(map(int, input().split()))

        if command == 'intersection_update':
            a.intersection_update(other)
        elif command == 'update':
            a.update(other)
        elif command == 'symmetric_difference_update':
            a.symmetric_difference_update(other)
        elif command == 'difference_update':
            a.difference_update(other)

    print(sum(a))
```

# The Captain's Room


```python
if __name__ == '__main__':
    k = int(input())
    rooms = list(map(int, input().split()))

    unique_rooms = set(rooms)

    captain_room = (sum(unique_rooms) * k - sum(rooms)) // (k - 1)

    print(captain_room)
```

# Check Subset


```python
if __name__ == '__main__':
    t = int(input())

    for _ in range(t):
        n = int(input())
        a = set(map(int, input().split()))

        m = int(input())
        b = set(map(int, input().split()))

        print(a.issubset(b))
```

# Check Strict Superset


```python
if __name__ == '__main__':
    a = set(map(int, input().split()))
    n = int(input())

    result = True

    for _ in range(n):
        other = set(map(int, input().split()))

        if not a > other:
            result = False
            break

    print(result)
```
