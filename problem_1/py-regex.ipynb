# Detect Floating Point Number

```python
import re

for _ in range(int(input())):
    print(bool(re.fullmatch(r'[+-]?[0-9]*\.[0-9]+', input())))
```

# Re.split()

```python
import re

print('\n'.join(re.split(r'[,.]', input())))
```

# Group(), Groups() & Groupdict()

```python
import re

s = input()
m = re.search(r'([a-zA-Z0-9])\1+', s)
print(m.group(1) if m else -1)
```

# Re.findall() & Re.finditer()

```python
import re

s = input()
matches = re.findall(r'(?<=[b-df-hj-np-tv-zB-DF-HJ-NP-TV-Z])[aeiouAEIOU]{2,}(?=[b-df-hj-np-tv-zB-DF-HJ-NP-TV-Z])', s)
print('\n'.join(matches) if matches else -1)
```

# Re.start() & Re.end()

```python
import re

s = input()
k = input()
matches = list(re.finditer(r'(?={})'.format(re.escape(k)), s))
if matches:
    for match in matches:
        print((match.start(), match.start() + len(k) - 1))
else:
    print((-1, -1))
```

# Regex Substitution

```python
import re

for _ in range(int(input())):
    line = input()
    line = re.sub(r'(?<= )&&(?= )', 'and', line)
    line = re.sub(r'(?<= )\|\|(?= )', 'or', line)
    print(line)
```

# Validating Roman Numerals

```python
import re

pattern = r'M{0,3}(?:CM|CD|D?C{0,3})(?:XC|XL|L?X{0,3})(?:IX|IV|V?I{0,3})'
print(bool(re.fullmatch(pattern, input())))
```

# Validating phone numbers

```python
import re

for _ in range(int(input())):
    print('YES' if re.fullmatch(r'[789][0-9]{9}', input()) else 'NO')
```

# Validating and Parsing Email Addresses

```python
import re
from email.utils import parseaddr

for _ in range(int(input())):
    line = input()
    name, address = parseaddr(line)
    email_pattern = r'[A-Za-z][A-Za-z0-9._-]*@[A-Za-z]+\.[A-Za-z]{1,3}'
    if name and re.fullmatch(email_pattern, address):
        print(line)
```

# Hex Color Code

```python
import re

for _ in range(int(input())):
    line = input()
    if '{' in line or '}' in line:
        print('\n'.join(re.findall(r'#[0-9a-fA-F]{3,6}(?=[;,)\s])', line)))
```
