# Detect Floating Point Number

```python
import re

for _ in range(int(input())):
    print(bool(re.fullmatch(r'[+-]?[0-9]*\.[0-9]+', input())))
```

# Group(), Groups() & Groupdict()

```python
import re

s = input()
m = re.search(r'([a-zA-Z0-9])\1+', s)
print(m.group(1) if m else -1)
```

# Re.findall() & Re.finditer()

```python
import re

s = input()
matches = re.findall(r'(?<=[b-df-hj-np-tv-zB-DF-HJ-NP-TV-Z])[aeiouAEIOU]{2,}(?=[b-df-hj-np-tv-zB-DF-HJ-NP-TV-Z])', s)
print('\n'.join(matches) if matches else -1)
```

# Re.start() & Re.end()

```python
import re

s = input()
k = input()
matches = list(re.finditer(r'(?={})'.format(re.escape(k)), s))
if matches:
    for match in matches:
        print((match.start(), match.start() + len(k) - 1))
else:
    print((-1, -1))
```

# Regex Substitution

```python
import re

for _ in range(int(input())):
    line = input()
    line = re.sub(r'(?<= )&&(?= )', 'and', line)
    line = re.sub(r'(?<= )\|\|(?= )', 'or', line)
    print(line)
```

# Validating and Parsing Email Addresses

```python
import re
from email.utils import parseaddr

for _ in range(int(input())):
    line = input()
    name, address = parseaddr(line)
    email_pattern = r'[A-Za-z][A-Za-z0-9._-]*@[A-Za-z]+\.[A-Za-z]{1,3}'
    if name and re.fullmatch(email_pattern, address):
        print(line)
```

# Hex Color Code

```python
import re

n = int(input())
inside_block = False

for _ in range(n):
    line = input()

    if '{' in line:
        inside_block = True
        continue

    if '}' in line:
        inside_block = False
        continue

    if inside_block:
        colors = re.findall(r'#[0-9a-fA-F]{3}(?:[0-9a-fA-F]{3})?', line)
        for color in colors:
            print(color)
```

# HTML Parser - Part 1

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(f'Start : {tag}')
        for attr, value in attrs:
            print(f'-> {attr} > {value}')

    def handle_endtag(self, tag):
        print(f'End   : {tag}')

    def handle_startendtag(self, tag, attrs):
        print(f'Empty : {tag}')
        for attr, value in attrs:
            print(f'-> {attr} > {value}')

parser = MyHTMLParser()
parser.feed(''.join(input() + '\n' for _ in range(int(input()))))
```

# HTML Parser - Part 2

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, data):
        lines = data.split('\n')
        if len(lines) == 1:
            print('>>> Single-line Comment')
            print(data)
        else:
            print('>>> Multi-line Comment')
            print(data)

    def handle_data(self, data):
        if data.strip():
            print('>>> Data')
            print(data)

parser = MyHTMLParser()
parser.feed(''.join(input() + '\n' for _ in range(int(input()))))
```

# Detect HTML Tags, Attributes and Attribute Values

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(tag)
        for name, value in attrs:
            print(f'-> {name} > {value}')

    def handle_startendtag(self, tag, attrs):
        print(tag)
        for name, value in attrs:
            print(f'-> {name} > {value}')

parser = MyHTMLParser()
parser.feed(''.join(input() + '\n' for _ in range(int(input()))))
```

# Validating UID

```python
import re

for _ in range(int(input())):
    s = input()
    print('Valid' if (
        len(s) == 10
        and s.isalnum()
        and sum(c.isupper() for c in s) >= 2
        and sum(c.isdigit() for c in s) >= 3
        and len(set(s)) == 10
    ) else 'Invalid')
```

# Validating Credit Card Numbers

```python
import re

for _ in range(int(input())):
    s = input()
    if re.fullmatch(r'(?!.*(\d)(-?\1){3})[456]\d{3}(-?\d{4}){3}', s):
        print('Valid')
    else:
        print('Invalid')
```

# Validating Postal Codes

```python
regex_integer_in_range = r"^[1-9]\d{5}$"
regex_alternating_repetitive_digit_pair = r"(\d)(?=\d\1)"

import re

P = input()

print(
    bool(re.match(regex_integer_in_range, P))
    and len(re.findall(regex_alternating_repetitive_digit_pair, P)) < 2
)
```

# Matrix Script

```python
import re

n, m = map(int, input().rstrip().split())
matrix = [input() for _ in range(n)]

decoded = ''.join(matrix[i][j] for j in range(m) for i in range(n))

print(re.sub(r'(?<=\w)[^\w]+(?=\w)', ' ', decoded))
```
