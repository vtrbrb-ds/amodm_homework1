# List Comprehensions


```python
x = int(input())
y = int(input())
z = int(input())
n = int(input())

result = [
    [i, j, k]
    for i in range(x + 1)
    for j in range(y + 1)
    for k in range(z + 1)
    if i + j + k != n
]

print(result)
```

# Find the Runner-Up Score!


```python
n = int(input())
scores = list(map(int, input().split()))

unique_scores = sorted(set(scores))

print(unique_scores[-2])
```

# Nested Lists


```python
students = []

for _ in range(int(input())):
    name = input()
    grade = float(input())
    students.append([name, grade])

grades = sorted(set(student[1] for student in students))
second_lowest = grades[1]

names = sorted(
    student[0]
    for student in students
    if student[1] == second_lowest
)

for name in names:
    print(name)
```

# Finding the Percentage


```python
n = int(input())

students = {}

for _ in range(n):
    name, *marks = input().split()
    students[name] = list(map(float, marks))

query_name = input()

average = sum(students[query_name]) / len(students[query_name])

print(f"{average:.2f}")
```

# Lists


```python
if __name__ == '__main__':
    N = int(input())
    numbers = []

    for _ in range(N):
        command = input().split()
        operation = command[0]

        if operation == "insert":
            numbers.insert(int(command[1]), int(command[2]))
        elif operation == "print":
            print(numbers)
        elif operation == "remove":
            numbers.remove(int(command[1]))
        elif operation == "append":
            numbers.append(int(command[1]))
        elif operation == "sort":
            numbers.sort()
        elif operation == "pop":
            numbers.pop()
        elif operation == "reverse":
            numbers.reverse()
```
