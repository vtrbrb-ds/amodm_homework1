Zipped

```python
n, x = map(int, input().split())

marks = [list(map(float, input().split())) for _ in range(x)]

for student_marks in zip(*marks):
    print(sum(student_marks) / x)
```

```python
s = input()
print(''.join(sorted(s, key=lambda c: (c.isdigit() - c.islower(), c in '02468', c))))
```
